# 01 — SAR & AIS Exploratory Data Analysis

This notebook is used for the initial exploratory data analysis (EDA) of the two core data sources in **dark-vessel-tracker**:

- **Sentinel-1 SAR imagery** — backscatter distributions, sea clutter vs. vessel signatures, scene coverage over the area of interest.
- **AIS position reports** — coverage, reporting gaps, vessel types and data-quality issues.

The findings here inform the detection thresholds and AIS matching parameters used in `src/processing/coordinate_matcher.py`.
